# Include collision kind and type

```python
# bring UK data to format kind_of_accident, make column for every kind of accident 
# 1 = Collision with another vehicle which starts, stops or is stationary 
# vehicle_maneuvre: 2 parked, 20 parking, 5: moving off, 14: Overtaking stationary vehicle on its offside

# 2 = Collision with another vehicle moving ahead or
# waiting 
# 19: going ahead 3: Waiting to go ahead 15:Overtaking on nearside (passengers side nearest kerb)

# 3 = Collision with another vehicle moving laterally
# in the same direction
# vehicle_maneuvre 13: Over taking moving vehicle on its offside

# 4 = Collision with another oncoming vehicle
# vehicle_direction_from vehicle_dirction_to:  0, 99, -1 not in entry

# 5 = Collision with another vehicle which turns into
# or crosses a road 
# vehicle_maneuvre: 6: U-turn 7: turning left, 9: turning right, 8: waiting to turn left, 10: waiting to turn right

# 6 = Collision between vehicle and pedestrian
# if is_pedestrian == 1 and one of the other

# 7 = Collision with an obstacle in the carriageway
# hit_object_in_carriageway: 1-12 

# 8 = Leaving the carriageway to the right
# vehicle_leaving_carriageway: 4-8 offside else

# 9 = Leaving the carriageway to the left
# vehicle_leaving_carriageway: 1,2 Nearside

# 0 = Accident of another kind
# vehicle_maneuvre: 99, -1 in entry

merged_df_uk_all['vehicle_manoeuvre'].head(20)
```

```python
# reduce vehicle types to german format

# 1 = Collision with another vehicle which starts, stops or is stationary 
# vehicle_maneuvre: 2 parked,4: Slowing or stopping,5: moving off , 14: Overtaking stationary vehicle on its offside ,20 parking
kind_of_accident_1 = [2,4,5,14,20]

# 2 = Collision with another vehicle moving ahead or waiting 
# 3: Waiting to go ahead 15:Overtaking on nearside (passengers side nearest kerb) 19: going ahead 
kind_of_accident_2 = [3,8,10,15,19]

# 3 = Collision with another vehicle moving laterally
# in the same direction
# vehicle_maneuvre 13: Over taking moving vehicle on its offside
kind_of_accident_3 = [11,12,13]

# 4 = Collision with another oncoming vehicle
# vehicle_direction_from vehicle_dirction_to:  0, 99, -1 not in entry
kind_of_accident_4 = [0,99,-1]

# 5 = Collision with another vehicle which turns into
# or crosses a road 
# vehicle_maneuvre: 6: U-turn 7: turning left, 9: turning right, 8: waiting to turn left, 10: waiting to turn right
kind_of_accident_5 = [6,7,9,8,10]

# 6 = Collision between vehicle and pedestrian
# if is_pedestrian == 1 and one of the other


# 7 = Collision with an obstacle in the carriageway
# hit_object_in_carriageway: 1-12, not in 0,99,-1
kind_of_accident_7 = [0,99,-1]


# 0 = Accident of another kind
# vehicle_maneuvre: 99, -1 in entry
kind_of_accident_0 = [99,-1]

# casualties
def extract_flags(vehicle_maneuvre):
    if not isinstance(vehicle_maneuvre, list):
        vehicle_maneuvre = [vehicle_maneuvre]
        
    # initialize flags
    flags = {
        "kind_of_accident_1": 0,
        "kind_of_accident_2": 0,
        "kind_of_accident_3": 0,
        "kind_of_accident_5": 0,
        "kind_of_accident_7": 0,
        "kind_of_accident_0": 0,
    }

    for v in vehicle_maneuvre:
        # mapping
        
        if v in kind_of_accident_1:
            flags["kind_of_accident_1"] = 1

        if v in kind_of_accident_2:
            flags["kind_of_accident_2"] = 1

        if v in kind_of_accident_3:
            flags["kind_of_accident_3"] = 1

        if v in kind_of_accident_7:
            flags["kind_of_accident_7"] = 1

        if v in kind_of_accident_0:
            flags["kind_of_accident_0"] = 1
    
    if len(vehicle_maneuvre) >=2:
        for v in vehicle_maneuvre:
            if v in kind_of_accident_5:
                flags["kind_of_accident_5"] = 1



    return pd.Series(flags)


# Flags for whole dataset
merged_df_uk_all_flags = merged_df_uk_all['vehicle_manoeuvre'].apply(extract_flags)


merged_df_uk_all = pd.concat([merged_df_uk_all, merged_df_uk_all_flags], axis=1)

merged_df_uk_all




```

```python
# add carriageways
# 8 = Leaving the carriageway to the right
# vehicle_leaving_carriageway: 4-8 offside else
kind_of_accident_8 = [4,5,6,7,8]

# 9 = Leaving the carriageway to the left
# vehicle_leaving_carriageway: 1,2 Nearside
kind_of_accident_9 = [1,2]


def extract_flags(vehicle_leaving_carr):
    if not isinstance(vehicle_leaving_carr, list):
        vehicle_leaving_carr = [vehicle_leaving_carr]

    # initialize flags
    flags = {
            "kind_of_accident_8": 0,
            "kind_of_accident_9": 0,
        }
    for v in vehicle_leaving_carr:
            # mapping
        if v in kind_of_accident_8:
            flags["kind_of_accident_8"] = 1

        if v in kind_of_accident_9:
            flags["kind_of_accident_9"] = 1

        
    return pd.Series(flags)


# Flags for whole dataset

merged_df_uk_all_flags = merged_df_uk_all['vehicle_leaving_carriageway'].apply(extract_flags)

merged_df_uk_all = pd.concat([merged_df_uk_all, merged_df_uk_all_flags], axis=1)
merged_df_uk_all
```

```python
# 6 = Collision between vehicle and pedestrian
# if is_pedestrian == 1 and one of the other
# cols = ['is_car','is_cyle','is_mcyle','is_goodsv','is_other']

cols = ['is_car','is_cyle','is_mcyle','is_goodsv','is_other']
merged_df_uk_all['kind_of_accident_6'] = 0 
print(merged_df_uk_all['is_pedestrian'])
merged_df_uk_all.loc[
    (merged_df_uk_all['is_pedestrian'] == 1) & 
    (merged_df_uk_all[cols].sum(axis=1) >= 1), 
    'kind_of_accident_6'
] = 1


# merged_df_uk_all
```

```python
# 4 = Collision with another oncoming vehicle
# vehicle_direction_from vehicle_dirction_to:  0, 99, -1 not in entry
kind_of_accident_4 = [0,99,-1]


def extract_flags(directions):
    dir_from,dir_to = directions
    if not isinstance(dir_from, list):
        dir_from = [dir_from]
    if not isinstance(dir_to, list):
        dir_to = [dir_to]
    

    # initialize flags
    flags = {
            "kind_of_accident_4": 0,
        }
    
    if len(dir_from) > 1 or len(dir_to) >1:
       if set(dir_from).isdisjoint(kind_of_accident_4) and set(dir_to).isdisjoint(kind_of_accident_4):
            flags["kind_of_accident_4"] = 1
                #print(9)
            
    return pd.Series(flags)

# Flags for whole dataset
merged_df_uk_all_flags = merged_df_uk_all[['vehicle_direction_from','vehicle_direction_to']].apply(lambda x: extract_flags((x['vehicle_direction_from'], x['vehicle_direction_to'])), axis=1)

merged_df_uk_all = pd.concat([merged_df_uk_all, merged_df_uk_all_flags], axis=1)

merged_df_uk_all
```

```python
# brind UK data to format type_of_accident
# 1 = Driving accident
# accidents without pedestrians
merged_df_uk_all['type_of_accident_1'] = merged_df_uk_all['is_pedestrian'].apply(lambda x: 1 if x == 0 else 0)

# 2 = Accident caused by turning off the road
# vehicle_manoeuvre: 7: Turning left, 9: Turning right
merged_df_uk_all['type_of_accident_2'] = merged_df_uk_all['vehicle_manoeuvre'].apply(lambda x: 1 if ((x == 1) or (x == 2) or (x == 3)or (x == 4)) else 0)

# 3 = Accident caused by turning into a road or by crossing it
# junction_location: 2: Cleared junction or waiting/parked at junction exit, 3: Leaving roundabout, 4: Entering roundabout,
# 5: Leaving main road, 6:Entering main road 7:Entering from slip road 8:Mid Junction - on roundabout or on main road
merged_df_uk_all['type_of_accident_3'] = merged_df_uk_all['junction_location'].apply(lambda x: 1 if ((x == 2) or (x == 3) or (x == 4)or (x == 5) or (x == 6) or (x == 7)or (x == 8)) else 0)


# 4 = Accident caused by crossing the road
# pedestrian_movement: 1:Crossing from driver's nearside ,2: Crossing from nearside - masked by parked or stationary vehicle ,3: Crossing from driver's offside,4:Crossing from offside - masked by  parked or stationary vehicle
merged_df_uk_all['type_of_accident_4'] = merged_df_uk_all['pedestrian_movement'].apply(lambda x: 1 if ((x == 7) or (x == 9)) else 0)


# 5 = Accident involving stationary
# vehicle_manoeuvre: 3: Waiting to go ahead, 4: Slowing or stopping
merged_df_uk_all['type_of_accident_5'] = merged_df_uk_all['vehicle_manoeuvre'].apply(lambda x: 1 if ((x == 3) or (x == 4)) else 0)

# 6 = Accident between vehicles moving along in carriageway
# vehicle_manoeuvre: 11: Changing lane to left 12: Changing lane to right, 13: Over taking moving vehicle on its offside
merged_df_uk_all['type_of_accident_6'] = merged_df_uk_all['vehicle_manoeuvre'].apply(lambda x: 1 if ((x == 11) or (x == 12)or (x == 13)) else 0)

# 7 = Other accident
# not 
```
