```python
#make entry to list if one accident appears more than on in df
def to_list_if_needed(x): 
    uniq = pd.unique(x)
    if len(uniq) == 1:
        return uniq[0]
    else:
        return uniq.tolist()

# one entry for every collision
df_vehicles_uk_clean = df_vehicles_uk.groupby('collision_index').agg(to_list_if_needed).reset_index()
df_vehicles_uk_clean
```

```python
# one entry for every collision
df_casualties_uk_clean = df_casualties_uk.groupby('collision_index').agg(to_list_if_needed).reset_index()
df_casualties_uk_clean
```

# Now make UK data more similar with Germany with the 

## Vehicle types

```python
# reduce vehicle types to german format

mcyle_codes = [2,3,4,5,23,97]

car_codes = [8, 9, 10, 19]

goodsv_codes = [11,20,21,98]

other_codes = [16,17,18,22,90,99]


# casualties
def extract_flags(vehicle_list):
    if not isinstance(vehicle_list, list):
        vehicle_list = [vehicle_list]

    # initialize flags
    flags = {
        "is_car": 0,
        "is_cyle": 0,
        "is_mcyle": 0,
        "is_goodsv": 0,
        "is_other": 0,
    }

    for v in vehicle_list:
        # mapping
        if isinstance(v, list):
            if 1 in v:
                flags["is_cyle"] = 1
        else:
            if v == 1:
                flags["is_cyle"] = 1
            
        if v in mcyle_codes:
            flags["is_mcyle"] = 1

        if v in car_codes:
            flags["is_car"] = 1

        if v in goodsv_codes:
            flags["is_goodsv"] = 1

        if v in other_codes:
            flags["is_other"] = 1
    
    return pd.Series(flags)


# Flags for whole dataset
merged_df_uk_all_flags = merged_df_uk_all['vehicle_type'].apply(extract_flags)


merged_df_uk_all = pd.concat([merged_df_uk_all, merged_df_uk_all_flags], axis=1)

merged_df_uk_all




```

## Pedestrians

```python
def extract_flags(casualty_type_list):
    if not isinstance(casualty_type_list, list):
        casualty_type_list = [casualty_type_list]

    # initialize flags
    flags = {
            "is_pedestrian": 0,
        }
    for v in casualty_type_list:
            # mapping
        if v != 0:
            flags["is_pedestrian"] = 1
        
    return pd.Series(flags)


# Flags for whole dataset

merged_df_uk_all_pedestrian = merged_df_uk_all['pedestrian_location'].apply(extract_flags)

merged_df_uk_all = pd.concat([merged_df_uk_all, merged_df_uk_all_pedestrian], axis=1)

merged_df_uk_all
```
