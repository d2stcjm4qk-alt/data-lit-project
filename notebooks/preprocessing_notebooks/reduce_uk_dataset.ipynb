## Add urban_or_rural_area column to the german dataset

```python
# collisions urban area germany
df_urban_rural_ger = pd.read_csv('../../data/preprocessed/germany/collisions/intermediate_steps/collisions_urban_area_ger.csv',low_memory=False, dtype={'c_id': str})
# german collision dataset 
#df_collisions_ger = pd.read_csv('../data/processed/modified_ger.csv',low_memory=False, dtype={'c_id': str})
print(df_urban_rural_ger['c_id'].dtypes)
print(df_collisions_ger['c_id'].dtypes)
```

```python

df_collisions_ger['urban_or_rural_area'] = df_collisions_ger['c_id'].isin(df_urban_rural_ger['c_id']).map({True: 1, False: 2})
df_collisions_ger
```
